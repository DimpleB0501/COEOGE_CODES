```python
import cv2
import numpy as np
import time
from ultralytics import RTDETR
import torch

# --- Configuration ---
model_path = "/home/dimple/Desktop/Model_Training/Training_Results/RTDETR_200_1280/weights/best.pt"
image_path = "/home/dimple/Desktop/Model_Training/Test_Images/IMG_0501_1_5_Red edge.tif"   
device = 0 if torch.cuda.is_available() else "cpu"

# --- Load model ---
model = RTDETR(model_path)

# --- Load and preprocess image ---
image = cv2.imread(image_path, cv2.IMREAD_UNCHANGED)
if image is None:
    raise FileNotFoundError(f"Image not found: {image_path}")

if image.dtype == np.uint16:
    image = (image / 256).astype(np.uint8)

if len(image.shape) == 2:
    image = cv2.cvtColor(image, cv2.COLOR_GRAY2RGB)
elif image.shape[2] == 1:
    image = cv2.cvtColor(image, cv2.COLOR_GRAY2RGB)

# --- Warm-up (optional but helps GPU consistency) ---
_ = model.predict(source=image, device=device, verbose=False)

# --- Time inference for 1 image ---
start = time.time()
_ = model.predict(source=image, device=device, verbose=False)
end = time.time()

# --- Report result ---
print(f"\n Inference time for 1 image using RT-DETR: {end - start:.4f} seconds")
```
