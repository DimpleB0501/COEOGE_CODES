```python
import cv2
import numpy as np
import time
from ultralytics import YOLO

# --- Load model ---
model_path = "/home/dimple/Desktop/Model_Training/Training_Results/Yolov11M_500_1024/weights/best.pt"
model = YOLO(model_path)
device = 0  # Use "cpu" if no GPU

# --- Load and preprocess image ---
image_path = "/home/dimple/Desktop/Model_Training/Test_Images/IMG_0501_1_5_Red edge.tif"  # 🔁 Replace with your actual image path
image = cv2.imread(image_path, cv2.IMREAD_UNCHANGED)

# Convert to uint8 if needed
if image is None:
    raise FileNotFoundError(f"Image not found: {image_path}")

if image.dtype == np.uint16:
    print("Converting image from uint16 to uint8...")
    image = (image / 256).astype(np.uint8)

# --- Optional: convert grayscale to 3-channel ---
if len(image.shape) == 2:
    image = cv2.cvtColor(image, cv2.COLOR_GRAY2RGB)
elif image.shape[2] == 1:
    image = cv2.cvtColor(image, cv2.COLOR_GRAY2RGB)

# --- Warm-up prediction (important for timing accuracy) ---
_ = model.predict(source=image, device=device, verbose=False)

# --- Time the prediction ---
start_time = time.time()
results = model.predict(source=image, device=device, verbose=False)
end_time = time.time()

# --- Output ---
print(f"Prediction time for 1 image: {end_time - start_time:.4f} seconds")
```
